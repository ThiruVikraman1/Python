DICTIONARIES

1. Mutable

2. Stored in key value pairs

3. Ordered

4. Unique keys - you cannot have 2 values for same key the new value over writes the old value

```python
# Syntax
# Empty dictionary
my_dict = {}

# A dictionary representing an AC unit (Engineering Context)
ac_details = {"brand": "Daikin","tonnage": 1.5,"star_rating": 5,"inverter": True}
```

Operation,Syntax,Result

1.Accessing -  dic_name["key"]             - Returns the value of the key

2.Adding    -  dic_name["key"] = value     - Adds a new key value pair to the dictionary.

3.Updating  -  dic_name["key"] = new value - Changes the value of an existing key.

4.Removing  -  dic_name.pop("key")         - Removes the specified key and its value.


```python
# Accessing
ac_details["brand"]
```

```python
# Adding 
ac_details["price"] = 45000
print(ac_details)
```

```python
# Updating
ac_details["tonnage"] = 1.0
print(ac_details)
```

```python
# Removing
ac_details.pop("price")
print(ac_details)
```

```python
# You have a list of fruits sold in a day. Create a dictionary that counts how many of each fruit were sold.
sales = ["apple", "banana", "apple", "orange", "banana", "apple"]

counts={}
for i in sales:
    if i in counts:
        counts[i]+=1
    else:
        counts[i] = 1
print(counts)
```

DICTIONARY METHODS

Method     -      Syntax           -    What it returns

.keys()    -    dic_name.keys()    -    A list of all unique fruits (keys).

.values()  -    dic_name.values()  -    A list of all the counts (values).

.items()   -    dic_name.items()   -    A list of tuples containing both fruit and count.

```python
print(counts)
```

```python
# .get method 
counts = {}
for i in sales:
    # .get(i, 0) looks for 'i'. If not found, it returns 0.
    counts[i] = counts.get(i, 0) + 1

print(counts)
```

```python
for fruit, total in counts.items():
    print ("total",fruit,"sold",total)
```

```python
#find which fruit has the maximum value
max_fruit = ""
max_count = 0
for fruit, total in counts.items():
    if total > max_count:
        max_count = total
        max_fruit = fruit

print("The most sold fruit is",max_fruit,max_count)
```
