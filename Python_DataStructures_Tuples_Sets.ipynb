TUPLES

1. Immutable - values can't be changes after creation

2. Ordered 

3. Allows duplicates

4. Allows all types of data - (HETEROGENEOUS)

```python
# syntax
# Empty tuple
empty_tuple = ()

int_tuple = (20,30)

# tuple with different types of data
het_tuple = ("Thiru",22,True)
```

```python
type(int_tuple)
```

TUPLE OPERATIONS

1. Accessing  - tuple_name[index]   - returns the value of the Index

2. Unpacking - This is a feature where you assign tuple values to variables in one line.

```python
# Accessing 
het_tuple[0]
```

```python
# Unpacking
name,age,correct = het_tuple
print(name,age)
```

SETS

1. Immutable - values can't be changes after creation

2. Unordered - values appear in different order every time we use them

3. Dose not allow duplicates - automatically removes repeating values
 

```python
# Syntax 
# Empty set
empty_set = set()
my_set = {"apple","banana","apple","watermelon"}  # Removes duplicates
print(my_set)
```

```python
type(my_set)
```

SET OPERATIONS

1. Adding   -  set_name.add(value)

2. Removing -  set_name.remove(value)

3. Math Operations  -  You can find the Union (all items from both sets) or Intersection (only items present in both)

```python
# Adding
my_set.add("Orange")
print(my_set)
```

```python
# Removing
my_set.remove("watermelon")
```

```python
print(my_set)
```

```python
# Union and Intersection
morning_batch = {"Anish", "Amara", "Arjun", "Charlie"}
evening_batch = {"Arjun", "David", "Amara", "Beatrice"}
Union_set = morning_batch.union(evening_batch)
intersection_set = morning_batch.intersection(evening_batch)
print("Union of both sets :",Union_set)
print("Intersection of both sets :",intersection_set)
```

```python
Registered_ids = {101, 102, 103, 104, 105}  # Sets are better than list in this case because sets uses hash tables
Current_User = 106
if Current_User in Registered_ids:
    print("Registered")
else:
    print("Access denied")
```

Task for both tuples and sets
Imagine you have a list of employee IDs that were recorded as they entered a building. Some IDs appear multiple times.
Task A: Use a Set to find exactly how many unique employees entered today.
Task B: You have a fixed "Company Name" and "Location" that should never change. Store these in a Tuple and use Unpacking to print them.

```python
# Task A
employee_log = [101, 104, 102, 101, 105, 104, 103, 101, 102, 106, 105, 107]
new_employee_log = set(employee_log)
print("Number of unique employees entered today :",len(new_employee_log))
```

```python
# Task B
Company_Info = ("TechCorp Industries", "Chennai, India", 2020)
Company_name,Location,Year = Company_Info
print("Company Name:",Company_name)
print("Location:",Location)
```
